# 04. 최신 시장 검증 및 타 플랫폼 벤치마크

## 분석 목적

앞선 1·2차 EDA에서는 2021~2022년 Netflix Top 10 데이터를 기반으로
장기흥행 콘텐츠의 장르, 언어, 제작국가, 초기 성과 및 TMDB 메타데이터 특성을 분석했다.

이번 단계에서는 해당 결과가 최근 콘텐츠 시장에서도 어느 정도 유효한지 확인하기 위해
2026년 상반기 Netflix 공식 공개자료와 타 플랫폼의 게임 IP 기반 콘텐츠 사례를 비교한다.

단, 2026년 자료는 기존 Netflix Top 10 데이터와 수집 기준 및 지표가 다르므로
기존 long_hit 데이터와 직접 결합하지 않고 최신 시장 흐름을 확인하는 검증 자료로 활용한다.

```python
# 2026 최신 Netflix Global Weekly 데이터를 불러와 기본 구조와 기간을 확인

import pandas as pd

global_weekly_2026 = pd.read_excel(
    "../DATA/RAW/[엔터]Netflix_Top10_2026_H1/global_weekly_2026-09-03.xlsx"
)

print("데이터 크기:", global_weekly_2026.shape)
print("\n컬럼:")
print(global_weekly_2026.columns.tolist())

print("\n기간:")
print(global_weekly_2026["week"].min(), "~", global_weekly_2026["week"].max())

display(global_weekly_2026.head())
```

```python
# Global Weekly 전체 데이터에서 2026년 상반기 데이터만 추출

global_weekly_2026["week"] = pd.to_datetime(
    global_weekly_2026["week"]
)

global_h1_2026 = global_weekly_2026[
    (global_weekly_2026["week"] >= "2026-01-01")
    & (global_weekly_2026["week"] <= "2026-06-30")
].copy()

print("2026 H1 데이터 크기:", global_h1_2026.shape)

print(
    "2026 H1 기간:",
    global_h1_2026["week"].min(),
    "~",
    global_h1_2026["week"].max()
)

print("\n카테고리:")
print(global_h1_2026["category"].value_counts())

display(global_h1_2026.head())
```

```python
# 2026년 상반기 Global Weekly 데이터에서 TV 콘텐츠만 추출

tv_h1_2026 = global_h1_2026[
    global_h1_2026["category"].str.contains("TV")
].copy()

print("2026 H1 TV 데이터 크기:", tv_h1_2026.shape)

print("\n카테고리:")
print(tv_h1_2026["category"].value_counts())

print("\n작품 수:")
print(tv_h1_2026["show_title"].nunique())

display(tv_h1_2026.head())
```

```python
# 2026년 상반기 TV 콘텐츠별 Top 10 최대 유지 주차를 확인

tv_2026_duration = (
    tv_h1_2026.groupby(
        ["show_title", "category"]
    )["cumulative_weeks_in_top_10"]
    .max()
    .reset_index()
    .rename(
        columns={
            "cumulative_weeks_in_top_10": "max_cumulative_weeks"
        }
    )
    .sort_values(
        "max_cumulative_weeks",
        ascending=False
    )
)

print("TV 콘텐츠 수:", len(tv_2026_duration))

display(tv_2026_duration.head(20))
```

```python
# 2026년 상반기 동안 각 TV 콘텐츠가 실제로 Top 10에 등장한 주 수를 계산

tv_2026_observed_weeks = (
    tv_h1_2026.groupby(
        ["show_title", "category"]
    )["week"]
    .nunique()
    .reset_index()
    .rename(
        columns={
            "week": "observed_weeks_2026_h1"
        }
    )
    .sort_values(
        "observed_weeks_2026_h1",
        ascending=False
    )
)

display(tv_2026_observed_weeks.head(20))
```

```python
# 2026년 상반기 TV 콘텐츠의 실제 Top 10 등장 주 수 분포를 확인

print(
    tv_2026_observed_weeks["observed_weeks_2026_h1"].describe()
)

display(
    tv_2026_observed_weeks.head(20)
)
```

```python
# 기존 2021~22 분석과 동일하게 TV 콘텐츠를 시즌·파트 단위 content_key로 구분

tv_h1_2026["content_key"] = tv_h1_2026["season_title"].fillna(
    tv_h1_2026["show_title"]
)

print("show_title 기준 작품 수:", tv_h1_2026["show_title"].nunique())
print("content_key 기준 콘텐츠 수:", tv_h1_2026["content_key"].nunique())

display(
    tv_h1_2026[
        ["show_title", "season_title", "content_key"]
    ].drop_duplicates().head(20)
)
```

```python
# 2026년 상반기 TV 콘텐츠의 시즌·파트별 실제 Top 10 등장 주 수를 계산

tv_2026_observed_weeks = (
    tv_h1_2026.groupby(
        ["content_key", "show_title", "category"]
    )["week"]
    .nunique()
    .reset_index()
    .rename(columns={
        "week": "observed_weeks_2026_h1"
    })
    .sort_values(
        "observed_weeks_2026_h1",
        ascending=False
    )
)

print(
    tv_2026_observed_weeks[
        "observed_weeks_2026_h1"
    ].describe()
)

display(tv_2026_observed_weeks.head(20))
```

```python
# 2026 H1 시즌·파트 단위에서 5주 이상 Top 10에 등장한 콘텐츠를 장기 유지 후보로 확인

tv_2026_observed_weeks["long_stay_2026"] = (
    tv_2026_observed_weeks["observed_weeks_2026_h1"] >= 5
).astype(int)

print(tv_2026_observed_weeks["long_stay_2026"].value_counts())
print(
    "장기 유지 비율:",
    round(tv_2026_observed_weeks["long_stay_2026"].mean() * 100, 2),
    "%"
)

display(
    tv_2026_observed_weeks[
        tv_2026_observed_weeks["long_stay_2026"] == 1
    ]
)
```

```python
# 2026 H1 장기 유지 여부별 English / Non-English 비중을 비교

language_group_2026 = (
    tv_2026_observed_weeks.groupby(
        ["category", "long_stay_2026"]
    )
    .size()
    .reset_index(name="content_count")
)

language_group_2026
```

```python
# 2026 H1 English / Non-English TV 콘텐츠의 장기 유지 비율을 계산

category_long_stay_summary = (
    tv_2026_observed_weeks
    .groupby("category")["long_stay_2026"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

category_long_stay_summary["long_stay_rate"] = (
    category_long_stay_summary["mean"] * 100
)

category_long_stay_summary = category_long_stay_summary[
    ["category", "count", "sum", "long_stay_rate"]
]

category_long_stay_summary
```

```python
# 2021~2022 Netflix TV 분석 데이터를 불러와 최신 데이터와 비교할 준비

netflix_tv_df = pd.read_csv(
    "../DATA/PROCESSED/netflix_tv_analysis.csv"
)

print("2021~2022 TV 데이터 크기:", netflix_tv_df.shape)
print(netflix_tv_df.columns.tolist())

display(netflix_tv_df.head())
```

```python
# 2021~2022 TV 데이터의 English / Non-English 장기흥행 비율을 계산

category_long_hit_2021 = (
    netflix_tv_df
    .groupby("category")["long_hit"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

category_long_hit_2021["long_hit_rate"] = (
    category_long_hit_2021["mean"] * 100
)

category_long_hit_2021 = category_long_hit_2021[
    ["category", "count", "sum", "long_hit_rate"]
]

category_long_hit_2021
```

```python
# 2021~2022와 2026 H1의 English / Non-English 장기 유지 경향을 한 표로 비교

comparison_category = pd.DataFrame({
    "category": ["TV (English)", "TV (Non-English)"],
    "2021_22_rate": [26.37, 29.41],
    "2026_h1_rate": [6.45, 12.00]
})

comparison_category
```

### English / Non-English 장기 유지 경향 비교

2021~2022년 분석에서는 TV (English)의 장기흥행 비율이 26.37%,
TV (Non-English)는 29.41%로 비영어권 콘텐츠가 조금 더 높게 나타났다.

2026년 상반기에는 5주 이상 Global Top 10에 등장한 콘텐츠를 장기 유지 후보로 정의했으며,
TV (English)는 6.45%, TV (Non-English)는 12.00%로 나타났다.

두 시기의 장기 유지 기준이 동일하지 않기 때문에 비율의 절대값을 직접 비교할 수는 없지만,
두 시기 모두 비영어권 TV 콘텐츠에서 상대적으로 높은 장기 유지 경향이 확인되었다.

따라서 두 시기의 장기 유지 기준은 다르지만,
비영어권 TV 콘텐츠가 영어권보다 상대적으로 높은 장기 유지 비율을 보였다는 점에서
유사한 방향성이 관찰되었다.
