```python
import pandas as pd

netflix = pd.read_csv(
    "../DATA/RAW/[엔터] Netflix Top 10 Weekly Dataset/all-weeks-global.csv"
)

netflix.head()
```

```python
print(netflix.shape)
print(netflix.columns)
```

```python
netflix[["show_title", "category"]].head(10)
```

```python
content_list = netflix[["show_title", "category"]].drop_duplicates()

content_list.head()
```

```python
content_list["tmdb_type"] = content_list["category"].apply(
    lambda x: "movie" if "Films" in x else "tv"
)

content_list.head(10)
```

```python
print(content_list.shape)
print(content_list["tmdb_type"].value_counts())
```

```python
# 작품 제목과 콘텐츠 타입(movie/tv)을 받아 TMDB 검색 결과를 반환하는 함수

def search_tmdb(title, tmdb_type):
    url = f"https://api.themoviedb.org/3/search/{tmdb_type}"

    params = {
        "api_key": TMDB_API_KEY,
        "query": title,
        "language": "en-US"
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        return response.json()["results"]
    else:
        return []
```

```python
# Netflix 작품 목록의 앞 5개를 TMDB에서 검색해 제목과 TMDB ID를 확인

for _, row in content_list.head(5).iterrows():

    results = search_tmdb(
        row["show_title"],
        row["tmdb_type"]
    )

    print("Netflix 제목:", row["show_title"])
    print("TMDB 타입:", row["tmdb_type"])

    if len(results) > 0:
        print("검색 1순위:", results[0].get("title") or results[0].get("name"))
        print("TMDB ID:", results[0].get("id"))
    else:
        print("검색 결과 없음")

    print("-" * 50)
```

```python
# Netflix 작품 전체를 TMDB에서 검색하고 1순위 후보의 기본 정보를 매칭 검증용 표로 저장

match_results = []

for _, row in content_list.iterrows():
    results = search_tmdb(
        row["show_title"],
        row["tmdb_type"]
    )

    if len(results) > 0:
        first_result = results[0]

        tmdb_title = first_result.get("title") or first_result.get("name")
        tmdb_date = first_result.get("release_date") or first_result.get("first_air_date")

        match_results.append({
            "netflix_title": row["show_title"],
            "category": row["category"],
            "tmdb_type": row["tmdb_type"],
            "tmdb_id": first_result.get("id"),
            "tmdb_title": tmdb_title,
            "tmdb_date": tmdb_date,
            "origin_country": first_result.get("origin_country")
        })

    else:
        match_results.append({
            "netflix_title": row["show_title"],
            "category": row["category"],
            "tmdb_type": row["tmdb_type"],
            "tmdb_id": None,
            "tmdb_title": None,
            "tmdb_date": None,
            "origin_country": None
        })
```

```python
# TMDB 매칭 결과를 데이터프레임으로 변환하고 앞부분을 확인

match_df = pd.DataFrame(match_results)

match_df.head(10)
```

```python
# TMDB 검색 결과가 없어 tmdb_id가 비어 있는 작품을 확인

match_df[match_df["tmdb_id"].isna()]
```

```python
# TMDB ID와 콘텐츠 타입(movie/tv)을 이용해 작품 상세정보를 가져오는 함수

def get_tmdb_details(tmdb_id, tmdb_type):
    url = f"https://api.themoviedb.org/3/{tmdb_type}/{int(tmdb_id)}"

    params = {
        "api_key": TMDB_API_KEY,
        "language": "en-US"
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        return response.json()
    else:
        return None
```

```python
# 첫 번째 작품 Brazen의 상세정보가 정상적으로 조회되는지 테스트

test_detail = get_tmdb_details(
    match_df.iloc[0]["tmdb_id"],
    match_df.iloc[0]["tmdb_type"]
)

test_detail.keys()
```

```python
# Brazen 상세정보에서 2차 EDA에 사용할 주요 변수만 확인

print("제목:", test_detail.get("title"))
print("장르:", test_detail.get("genres"))
print("공개일:", test_detail.get("release_date"))
print("러닝타임:", test_detail.get("runtime"))
print("제작 국가:", test_detail.get("production_countries"))
print("원어:", test_detail.get("original_language"))
print("평점:", test_detail.get("vote_average"))
print("투표 수:", test_detail.get("vote_count"))
print("인기도:", test_detail.get("popularity"))
```

```python
# TMDB 상세정보에서 분석에 필요한 변수만 정리해 저장할 빈 리스트 생성

tmdb_metadata = []

for _, row in match_df.iterrows():
    detail = get_tmdb_details(
        row["tmdb_id"],
        row["tmdb_type"]
    )

    if detail is None:
        continue

    genres = ", ".join(
        [genre["name"] for genre in detail.get("genres", [])]
    )

    production_countries = ", ".join(
        [country["iso_3166_1"] for country in detail.get("production_countries", [])]
    )

    tmdb_metadata.append({
        "netflix_title": row["netflix_title"],
        "tmdb_id": int(row["tmdb_id"]),
        "tmdb_type": row["tmdb_type"],
        "genres": genres,
        "original_language": detail.get("original_language"),
        "production_countries": production_countries,
        "release_date": detail.get("release_date"),
        "first_air_date": detail.get("first_air_date"),
        "runtime": detail.get("runtime"),
        "episode_run_time": detail.get("episode_run_time"),
        "number_of_seasons": detail.get("number_of_seasons"),
        "number_of_episodes": detail.get("number_of_episodes"),
        "vote_average": detail.get("vote_average"),
        "vote_count": detail.get("vote_count"),
        "popularity": detail.get("popularity")
    })
```

```python
# 수집한 TMDB 상세정보를 데이터프레임으로 변환하고 크기와 앞부분을 확인

tmdb_df = pd.DataFrame(tmdb_metadata)

print(tmdb_df.shape)
tmdb_df.head()
```

```python
# TV 작품의 상세정보가 정상적으로 수집되었는지 앞 5개를 확인

tmdb_df[tmdb_df["tmdb_type"] == "tv"].head()
```

```python
# TMDB 상세정보의 컬럼별 결측치 개수를 확인

tmdb_df.isna().sum()
```

```python
# TV 작품 중 episode_run_time이 빈 리스트인 작품 수를 확인

empty_episode_runtime = tmdb_df[
    (tmdb_df["tmdb_type"] == "tv") &
    (tmdb_df["episode_run_time"].apply(lambda x: x == []))
]

print(empty_episode_runtime.shape[0])
```

```python
# movie/tv 유형별로 주요 날짜와 러닝타임 결측치를 확인해 구조상 예외 작품이 있는지 점검

print(
    tmdb_df.groupby("tmdb_type")[
        [
            "release_date",
            "first_air_date",
            "runtime",
            "number_of_seasons",
            "number_of_episodes"
        ]
    ].count()
)
```

```python
# 수집이 완료된 TMDB 메타데이터를 PROCESSED 폴더에 CSV로 저장

tmdb_df.to_csv(
    "../DATA/PROCESSED/tmdb_metadata.csv",
    index=False,
    encoding="utf-8-sig"
)
```
